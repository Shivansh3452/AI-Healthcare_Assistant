```python
# Cell 2: Imports and model load
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

MODEL_NAME = "microsoft/Phi-3-mini-4k-instruct"  # instruction-tuned, fits free Colab GPU

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16,
    device_map="auto"
)

chatbot = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=512,
    temperature=0.3,       # low temp = fewer hallucinations
    do_sample=True,
    top_p=0.9
)
```

```python
# Cell 3: Reusable generation function
def ask_model(prompt, max_new_tokens=400):
    formatted = f"<|user|>\n{prompt}<|end|>\n<|assistant|>\n"
    output = chatbot(formatted, max_new_tokens=max_new_tokens, return_full_text=False)
    return output[0]["generated_text"].strip()

# Cell 4: Interactive loop
def run_chatbot():
    print("MediAssist AI — type 'exit' to stop")
    while True:
        user_input = input("User: ")
        if user_input.lower() == "exit":
            break
        response = ask_model(user_input)
        print(f"Assistant:\n{response}\n")

run_chatbot()
```

```python
task2_prompt = """
You are a licensed medical educator writing for first-year nursing students.

Context: The student needs a clear, medically accurate overview of a disease
for an exam preparation guide.

Constraints:
- Use only well-established, evidence-based medical information
- Do not speculate or give personalized medical advice
- Keep each section under 80 words
- Do not diagnose the reader

Task: Explain diabetes.

Output strictly in this Markdown structure:

## Disease Overview
## Symptoms
## Risk Factors
## Prevention
## When to Consult a Doctor
"""

print(ask_model(task2_prompt, max_new_tokens=600))
```

```python
# Cell 7
task4_prompt = """
You are a clinical triage assistant.

Patient data:
- Fever: present
- Cough: present
- Oxygen level: 88%

Internally work through the clinical significance of each symptom step by step,
but do NOT show your internal reasoning in the output.

Output ONLY the following, in Markdown:

## Assessment Summary
## Urgency Level
## Recommended Action
"""

print(ask_model(task4_prompt, max_new_tokens=300))
```

```python
# Cell 8
import json, re

task5_prompt = """
You are a medical data extraction system. Analyze the patient case and return
ONLY valid JSON, with no explanation, no markdown fences, no extra text.

Patient case: Fever, cough, and oxygen level of 88%.

Return exactly this schema:
{
  "disease": "",
  "symptoms": [],
  "risk_level": "",
  "recommendation": ""
}
"""

raw_output = ask_model(task5_prompt, max_new_tokens=200)
print("Raw:", raw_output)

# Defensive parsing (models often add stray text despite instructions)
match = re.search(r"\{.*\}", raw_output, re.DOTALL)
if match:
    try:
        parsed = json.loads(match.group())
        print(json.dumps(parsed, indent=2))
    except json.JSONDecodeError:
        print("Model did not return valid JSON — needs a stricter prompt or a JSON-mode model.")
```

```python
# Cell 9: Load a second model to compare against
MODEL_NAME_2 = "Qwen/Qwen2.5-1.5B-Instruct"
tokenizer2 = AutoTokenizer.from_pretrained(MODEL_NAME_2)
model2 = AutoModelForCausalLM.from_pretrained(MODEL_NAME_2, torch_dtype=torch.float16, device_map="auto")
chatbot2 = pipeline("text-generation", model=model2, tokenizer=tokenizer2, max_new_tokens=400)
```

```python
# Cell 10: Improved prompt
task7_prompt = """
You are a clinical decision-support assistant, NOT a diagnostician.
You must never state a definitive diagnosis — only rank possibilities.

Patient symptoms: fever, fatigue, muscle aches, recent travel to a tropical region.

Rules:
- Base reasoning ONLY on the symptoms given
- Provide a ranked differential (most to least likely) with a one-line rationale each
- For each, state what test would confirm it
- End with a clear recommendation to seek in-person clinical testing

Output in Markdown with headers: ## Differential Diagnosis, ## Recommended Tests, ## Next Steps
"""

print(ask_model(task7_prompt, max_new_tokens=400))
```

```python
# Cell 11
ROLE_INSTRUCTIONS = {
    "doctor": "Use precise clinical terminology. Include differential diagnoses, relevant lab/test suggestions, and treatment considerations. Assume advanced medical knowledge.",
    "nurse": "Focus on patient monitoring, symptom tracking, medication administration notes, and when to escalate to a doctor. Use clinical but practical language.",
    "medical_student": "Explain the underlying pathophysiology and reasoning in a teaching tone, as if preparing the student for an exam. Include the 'why' behind each point.",
    "patient": "Use simple, non-technical language. Be reassuring but honest. Avoid alarming terminology. Focus on what symptoms mean and what to do next."
}

def role_based_prompt(role, question):
    if role not in ROLE_INSTRUCTIONS:
        raise ValueError(f"Unknown role: {role}")
    return f"""
You are MediAssist AI. The user's role is: {role}.

Response style for this role: {ROLE_INSTRUCTIONS[role]}

Question: {question}

Respond in Markdown with appropriate headers for the role.
"""

# Example usage — same underlying model, different behavior per role
for role in ["doctor", "nurse", "medical_student", "patient"]:
    print(f"=== ROLE: {role} ===")
    print(ask_model(role_based_prompt(role, "What does a fever of 101°F with cough mean?"), max_new_tokens=250))
    print()
```

```python
# Cell: run same prompt through both models
test_prompt = """
You are a clinical educator. Explain diabetes in Markdown with these sections:
## Disease Overview
## Symptoms
## Risk Factors
## Prevention
## When to Consult a Doctor
Keep each section under 80 words.
"""

def ask_with(pipe, prompt, max_new_tokens=500):
    formatted = f"<|user|>\n{prompt}<|end|>\n<|assistant|>\n"
    out = pipe(formatted, max_new_tokens=max_new_tokens, return_full_text=False)
    return out[0]["generated_text"].strip()

response_phi3 = ask_with(chatbot, test_prompt)
response_qwen = ask_with(chatbot2, test_prompt)

print("=== Phi-3-mini ===")
print(response_phi3)
print("\n=== Qwen2.5-1.5B ===")
print(response_qwen)
```
